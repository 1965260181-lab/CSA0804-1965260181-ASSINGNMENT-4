```python
def read_attendance(data):

    records = {}
    errors = []

    for filename, content in data.items():

        for line_number, line in enumerate(content.splitlines(), 1):

            parts = [x.strip() for x in line.split(",")]

            if len(parts) != 3 or parts[2] not in ("P", "A"):
                errors.append(
                    f"{filename}:{line_number}: invalid record"
                )
                continue

            records.setdefault(parts[0], []).append(
                (parts[1], parts[2])
            )

    return records, errors


def calculate(records):

    summary = {}
    consecutive = []

    for reg, days in records.items():

        present = sum(status == "P" for _, status in days)
        absent = sum(status == "A" for _, status in days)

        summary[reg] = (present, absent)

        run = 0

        for _, status in days:

            run = run + 1 if status == "A" else 0

            if run >= 3:
                consecutive.append(reg)
                break

    return summary, consecutive


def write_reports(summary, consecutive, errors):

    print("\nATTENDANCE SUMMARY")

    for reg, (present, absent) in summary.items():
        print(f"{reg}: Present={present}, Absent={absent}")

    print("\nThree Consecutive Absences:", consecutive)
    print("Errors:", errors)


if __name__ == "__main__":

    sample_data = {
        "day1.txt": """1965260181,2026-10-01,A
1965260181,2026-10-02,A
1965260181,2026-10-03,A
1965260182,2026-10-01,P
1965260182,2026-10-02,A
"""
    }

    records, errors = read_attendance(sample_data)

    summary, consecutive = calculate(records)

    write_reports(summary, consecutive, errors)

    print("\nMODULE AND PACKAGE ORGANISATIONS")
    print("1. Single Script")
    print("2. One Module with Main Script")
    print("3. Three Modules")
    print("4. Package with __init__.py")
```
