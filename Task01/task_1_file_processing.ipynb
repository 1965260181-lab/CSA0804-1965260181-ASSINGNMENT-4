```python
from io import StringIO


def process_mess_bills():

    sample_data = """1965260181,Kaviya,2500
1965260182,Keerthana,3000
1965260183,Ruksana,2800
1965260184,Bhavna,invalid
1965260185,Anu,-500
"""

    total = 0
    report_lines = []
    rejected_lines = []

    file = StringIO(sample_data)

    for line_number, raw_line in enumerate(file, start=1):

        line = raw_line.strip()

        if not line:
            continue

        try:
            parts = [part.strip() for part in line.split(",")]

            if len(parts) != 3:
                raise ValueError("Expected 3 fields")

            register_number, name, amount_text = parts

            if not register_number or not name:
                raise ValueError("Missing register number or name")

            amount = float(amount_text)

            if amount < 0:
                raise ValueError("Amount cannot be negative")

            total += amount
            report_lines.append(f"{name}: Rs.{amount:.2f}")

        except ValueError as error:
            rejected_lines.append(
                f"Line {line_number}: {line} - {error}"
            )

    file.close()

    print("MESS BILL REPORT")
    print("----------------")

    for record in report_lines:
        print(record)

    print("----------------")
    print(f"Total Amount: Rs.{total:.2f}")

    print("\nRejected Records:")

    if rejected_lines:
        for record in rejected_lines:
            print(record)
    else:
        print("No rejected records")


if __name__ == "__main__":
    process_mess_bills()
```
