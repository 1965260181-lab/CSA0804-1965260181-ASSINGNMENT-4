```python
from io import StringIO
from collections import defaultdict

def process_month():

    daily_files = {
        "day01.txt": """1965260181,2026-10-01,P
1965260182,2026-10-01,A
damaged-line
""",

        "day02.txt": """1965260181,2026-10-02,P
1965260182,2026-10-02,P
"""
    }

    counts = defaultdict(lambda: {"P": 0, "A": 0})
    errors = []

    for filename, data in daily_files.items():

        with StringIO(data) as source:

            for line_number, raw in enumerate(source, start=1):

                line = raw.strip()

                if not line:
                    continue

                parts = [part.strip() for part in line.split(",")]

                if len(parts) != 3:
                    errors.append(
                        f"{filename}:{line_number}: expected 3 fields"
                    )
                    continue

                reg_no, date, status = parts

                if not reg_no or not date or status not in ("P", "A"):
                    errors.append(
                        f"{filename}:{line_number}: invalid record"
                    )
                    continue

                counts[reg_no][status] += 1

    print("MONTHLY ATTENDANCE SUMMARY")

    for reg_no in sorted(counts):

        present = counts[reg_no]["P"]
        absent = counts[reg_no]["A"]

        total = present + absent
        percentage = present / total * 100

        print(
            f"{reg_no}: Present={present}, "
            f"Absent={absent}, Attendance={percentage:.2f}%"
        )

    print("\nDamaged Lines:")

    for error in errors:
        print(error)


if __name__ == "__main__":
    process_month()
```
