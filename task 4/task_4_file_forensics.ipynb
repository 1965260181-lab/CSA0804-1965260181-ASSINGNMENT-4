```python
from io import StringIO

def save_returns(books, log_file):

    for book in books:
        log_file.write(book.rstrip("\n") + "\n")


def process_day(day_file, log_file, error_file):

    try:
        books = [
            line.strip()
            for line in StringIO(day_file)
            if line.strip()
        ]

        save_returns(books, log_file)

    except FileNotFoundError as error:
        error_file.write(f"{error}\n")
        print("Missing file")

    except PermissionError as error:
        error_file.write(f"{error}\n")
        print("Permission denied")


if __name__ == "__main__":

    returns_log = StringIO()
    errors_log = StringIO()

    daily_data = {
        "day1.txt": "Python Basics\nNetworking\n",
        "day2.txt": "Cyber Security\n",
        "day3.txt": "Data Structures\nOperating Systems\n"
    }

    for filename, data in daily_data.items():
        process_day(data, returns_log, errors_log)

    try:
        missing_file = None

        if missing_file is None:
            raise FileNotFoundError("input/missing.txt")

    except FileNotFoundError as error:
        errors_log.write(f"input/missing.txt: {error}\n")
        print("Missing file: input/missing.txt")

    print("\nRETURN RECORDS")
    print(returns_log.getvalue())

    print("ERROR LOG")
    print(errors_log.getvalue())
```
